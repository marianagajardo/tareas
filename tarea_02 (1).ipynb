## Problema 1. Clasificador de temperatura
Imprime los números de -30 a 50 y los clasifica como "Muy frío", "Templado" o "Caluroso".

```python
# Problema 1: Clasificador de temperatura
# Recorremos los números desde -30 hasta 50.
# range(-30, 51) llega hasta 50 porque el último número (51) no se incluye.
for temperatura in range(-30, 51):
    # Si es menor que 0, hace mucho frío
    if temperatura < 0:
        print(temperatura, "Muy frío")
    # Si está entre 0 y 24 (ambos incluidos), es templado
    elif temperatura <= 24:
        print(temperatura, "Templado")
    # En cualquier otro caso (25 o más), es caluroso
    else:
        print(temperatura, "Caluroso")
```

## Problema 2. Números especiales del robot
Imprime los números del 1 al 100 divisibles por 5 con "Bip", y con "Bip Bip" si además son divisibles por 10.

```python
# Problema 2: Números especiales del robot
# Recorremos los números del 1 al 100 (range no incluye el 101).
for numero in range(1, 101):
    # El operador % entrega el resto de la división.
    # Si el resto de dividir por 5 es 0, el número es divisible por 5.
    if numero % 5 == 0:
        # Primero revisamos si también es divisible por 10,
        # porque todo número divisible por 10 también lo es por 5.
        if numero % 10 == 0:
            print(numero, "Bip Bip")
        else:
            print(numero, "Bip")
    # Los números que no son divisibles por 5 no se imprimen.
```

## Problema 3. Detector de puntaje gamer
Pide dos puntajes, los suma y muestra el nivel según el resultado.

```python
# Problema 3: Detector de puntaje gamer
# input() siempre entrega texto, así que lo convertimos a número entero con int().
puntaje_1 = int(input("Ingresa primer puntaje: "))
puntaje_2 = int(input("Ingresa segundo puntaje: "))

# Sumamos ambos puntajes
suma = puntaje_1 + puntaje_2

# Mostramos el nivel según el resultado de la suma
if suma < 50:
    print("Resultado: Nivel principiante")
elif suma < 80:
    # Si llegamos aquí, la suma ya es mayor o igual a 50 y menor que 80
    print("Resultado: Nivel intermedio")
else:
    # Si no se cumplió nada de lo anterior, la suma es 80 o más
    print("Resultado: Nivel pro")
```

## Problema 4. Calculador de perfil musical
Una función recibe la edad y si a la persona le gusta la música y bailar, y retorna su perfil.

```python
# Problema 4: Calculador de perfil musical

# Creamos una función que recibe los tres parámetros y retorna un mensaje.
# edad: número entero
# le_gusta_musica y le_gusta_bailar: valores booleanos (True o False)
def perfil_musical(edad, le_gusta_musica, le_gusta_bailar):
    # Caso 1: 18 años o más, le gusta la música y le gusta bailar
    if edad >= 18 and le_gusta_musica and le_gusta_bailar:
        return "Eres una estrella de la fiesta"
    # Caso 2: le gusta la música pero NO le gusta bailar
    elif le_gusta_musica and not le_gusta_bailar:
        return "Te gusta la música, pero no tanto bailar"
    # Caso 3: cualquier otra combinación
    else:
        return "Tienes tu propio estilo"


# Le pedimos los datos al usuario
edad = int(input("¿Cuántos años tienes? "))

# Las respuestas de sí/no se convierten en True o False.
# .strip() quita espacios y .lower() pasa todo a minúscula, para aceptar "Si", "SI", "si", etc.
respuesta_musica = input("¿Te gusta la música? (si/no): ").strip().lower()
respuesta_bailar = input("¿Te gusta bailar? (si/no): ").strip().lower()

le_gusta_musica = respuesta_musica == "si" or respuesta_musica == "sí"
le_gusta_bailar = respuesta_bailar == "si" or respuesta_bailar == "sí"

# Llamamos a la función y mostramos el resultado
print(perfil_musical(edad, le_gusta_musica, le_gusta_bailar))
```
